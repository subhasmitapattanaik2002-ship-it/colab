#Create a Student class that stores name and age.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


s1 = Student("kiran", 20)
s2 = Student("jitu", 22)

s1.display()
s2.display()
```

**Six examples of class method**.

# 1.Access Class Variable

```python
class Student:
    school = "ABC Public School"   # Class variable (shared by all objects)

    @classmethod
    def show_school(cls):          # cls means Student class
        print("School Name:", cls.school)
                                   # cls.school = Student.school

                                    # Calling class method using class name
Student.show_school()
```

2.Change class variable

```python
class Employee:
    company = "Google"

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company
                                   # Changes the class variable

print("Before:", Employee.company)

Employee.change_company("Microsoft")

print("After:", Employee.company)
```

3.**Count objects**

```python
class Car:
    total_cars = 0

    def __init__(self):
        Car.total_cars += 1

    @classmethod
    def show_total(cls):
        print("Total Cars:", cls.total_cars)

c1 = Car()
c2 = Car()
c3 = Car()

Car.show_total()
```

4.Alternative constructor

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))

s1 = Student.from_string("Rahul,20")

print(s1.name)
print(s1.age)
```

6.Increase Bonus for All Employees

```python
class Employee:
    bonus = 5000

    @classmethod
    def increase_bonus(cls, amount):
        cls.bonus += amount

print("Old Bonus:", Employee.bonus)

Employee.increase_bonus(2000)

print("New Bonus:", Employee.bonus)
```

3.six examples of stastic method


1.Add two number


```python
class Calculator:

    @staticmethod
    def add(a, b):
                    # a and b are normal parameters
        return a + b

result = Calculator.add(10, 20)

print("Addition:", result)
```

2.Check even or odd

```python
class Number:

    @staticmethod
    def check(num):
        if num % 2 == 0:
            print(num, "is Even")
        else:
            print(num, "is Odd")

Number.check(8)
Number.check(5)
```

3.Celsius to Fahrenheit

```python
class Temperature:

    @staticmethod
    def celsius_to_fahrenheit(c):
        return (c * 9/5) + 32

print(Temperature.celsius_to_fahrenheit(30))
```

4.**Check Voting Eligibility**

```python
class Voting:

    @staticmethod
    def eligible(age):
        if age >= 18:
            print("Eligible to Vote")
        else:
            print("Not Eligible")

Voting.eligible(20)
Voting.eligible(15)
```

5.**Find Maximum Number**

```python
class MathOperations:

    @staticmethod
    def maximum(a, b):
        if a > b:
            return a
        else:
            return b

print(MathOperations.maximum(15, 30))
```

**DIFFERENCE BETWEEN CLASSIC METHOD, STASTIC METHOD AND INSTANCE METHOD**

| **Instance Method**                                | **Class Method**                                                    | **Static Method**                                                             |
| -------------------------------------------------- | ------------------------------------------------------------------- | ----------------------------------------------------------------------------- |
| Uses `self` as the first parameter.                | Uses `cls` as the first parameter.                                  | Uses no `self` or `cls`.                                                      |
| No decorator is used.                              | Uses `@classmethod`.                                                | Uses `@staticmethod`.                                                         |
| Works with instance (object) data.                 | Works with class data.                                              | Does not work with instance or class data directly.                           |
| Can access instance variables and class variables. | Can access only class variables directly.                           | Cannot access instance or class variables directly.                           |
| Mainly used for object-specific operations.        | Mainly used for class-level operations or alternative constructors. | Mainly used for utility/helper functions such as calculations or validations. |
| Usually called using an object: `obj.method()`.    | Usually called using the class: `ClassName.method()`.               | Usually called using the class: `ClassName.method()`.                         |
